## Analisis Pola Anomali Magnetik

Perubahan paling tajam pada contoh ini terdapat di antara jarak 400–600 meter, ketika anomali berubah dari +8 nT menjadi −18 nT. Peralihan dari anomali positif ke negatif dan perubahan gradien yang jelas dapat mengindikasikan adanya batas lateral antara material dengan sifat kemagnetan berbeda. Salah satu kemungkinan geologinya adalah kontak litologi atau zona struktur, termasuk kemungkinan sesar. Namun, pola ini belum cukup untuk memastikan keberadaan sesar karena geometri dan kedalaman sumber magnetik juga dapat menghasilkan variasi anomali yang serupa.

Berdasarkan profil ilustratif, zona sekitar jarak 400–600 meter merupakan bagian yang paling layak diperiksa lebih lanjut karena memperlihatkan perubahan anomali yang paling tajam. ## Visualisasi Data Ilustratif

Sel kode berikut menampilkan profil anomali magnetik dan gradien antar titik untuk membantu melihat zona perubahan yang paling tajam. Seluruh angka merupakan data simulasi, bukan data lapangan.

```python
import numpy as np
import matplotlib.pyplot as plt

# Data ilustratif/simulasi, bukan hasil pengukuran lapangan
jarak_m = np.arange(0, 1001, 100)
anomali_nT = np.array([12, 18, 25, 20, 8, -5, -18, -14, -4, 9, 15])

# Gradien anomali antar titik (nT/m)
gradien = np.diff(anomali_nT) / np.diff(jarak_m)
titik_tengah_m = (jarak_m[:-1] + jarak_m[1:]) / 2

plt.figure(figsize=(9, 4.8))
plt.plot(jarak_m, anomali_nT, marker="o")
plt.axvspan(400, 600, alpha=0.2, label="Zona perubahan tajam (400–600 m)")
plt.axhline(0, linewidth=1)
plt.xlabel("Jarak sepanjang lintasan (m)")
plt.ylabel("Anomali magnetik residual (nT)")
plt.title("Profil Anomali Magnetik Ilustratif")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(9, 4.2))
plt.plot(titik_tengah_m, gradien, marker="o")
plt.axvspan(400, 600, alpha=0.2, label="Zona perubahan tajam (400–600 m)")
plt.axhline(0, linewidth=1)
plt.xlabel("Jarak tengah antar titik (m)")
plt.ylabel("Gradien anomali (nT/m)")
plt.title("Gradien Anomali Magnetik Ilustratif")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print(f"Perubahan anomali dari 400 m ke 600 m: {anomali_nT[6] - anomali_nT[4]:.1f} nT")
print(f"Gradien rata-rata pada interval 400–600 m: {(anomali_nT[6] - anomali_nT[4]) / 200:.3f} nT/m")
```
